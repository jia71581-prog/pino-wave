# PINO HDF5 Acoustic FNO

Thin notebook wrapper around the modular `fno_acoustic` package. The original `Fourier_Acoustic_train.ipynb` is preserved unchanged.

```python
from pathlib import Path
import sys

ROOT = Path('..').resolve()
sys.path.insert(0, str(ROOT / 'src'))

from fno_acoustic.config import load_config
from fno_acoustic.train import run_training
from fno_acoustic.evaluate import run_evaluation

config_path = ROOT / 'configs' / 'pino_hdf5_smoke.yaml'
config = load_config(config_path)
config['data']['path']
```

```python
# Short smoke run only. Full baseline training should use configs/pino_hdf5.yaml from the CLI.
result = run_training(config_path, output_metrics=ROOT / 'artifacts/pino_adaptation/notebook_smoke_metrics.json')
result
```

```python
metrics = run_evaluation(config_path, result['checkpoint_best'], split='val')
metrics['mean']
```
